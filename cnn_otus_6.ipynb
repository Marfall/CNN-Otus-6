```python
# Ячейка 1: Загрузка датасета через Google Drive (gdown)
!pip install -q gdown

import gdown
import os
import zipfile

# ID файла из ссылки ТЗ
file_id = '1leoj71B_D_paBLC1uP5EKcnR5lJFPovn'
output_zip = 'cloud_dataset.zip'

# Скачивание, если файла ещё нет
if not os.path.exists(output_zip):
    print('Скачивание архива...')
    gdown.download(f'https://drive.google.com/uc?id={file_id}', output_zip, quiet=False)
else:
    print(f'Файл {output_zip} уже существует.')

# Распаковка
extract_dir = 'cloud_data'
if os.path.exists(output_zip):
    os.makedirs(extract_dir, exist_ok=True)
    with zipfile.ZipFile(output_zip, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)
    print(f'Архив распакован в папку: {extract_dir}')
    # Покажем верхнеуровневую структуру
    for root, dirs, files in os.walk(extract_dir):
        level = root.replace(extract_dir, '').count(os.sep)
        if level < 2:
            indent = ' ' * 4 * level
            print(f'{indent}{os.path.basename(root)}/')
            if level == 0:
                for f in files[:10]:
                    print(f'{indent}    {f}')
else:
    print('Ошибка: архив не найден. Проверь ссылку.')
```
